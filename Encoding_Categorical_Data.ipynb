```python
import numpy as np
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder(categories=[["small", "medium", "large"]]) # categories gives order 

sizes = ["small", "medium", "large"]
# reshape to 2D array
sizes = np.array(sizes).reshape(-1,1)
print(sizes)

encoded = encoder.fit_transform(sizes) 

print(encoded)
```

```python
import pandas as pd

# Sample data
data = {'Color': ['Red', 'Blue', 'Green', 'Blue', 'Red']}
df = pd.DataFrame(data)

# Applying One-Hot Encoding
df_encoded = pd.get_dummies(df, columns=['Color'])

print("Original DataFrame:")
print(df)

print("\nOne-Hot Encoded DataFrame:")
print(df_encoded)
```

```python
# Remove dummy variable trap using drop_first=True
import pandas as pd

# Sample data
data = {'Color': ['Red', 'Blue', 'Green', 'Blue', 'Red']}
df = pd.DataFrame(data)

# Applying One-Hot Encoding
df_encoded = pd.get_dummies(df, columns=['Color'],drop_first=True)

print("Original DataFrame:")
print(df)

print("\nOne-Hot Encoded DataFrame:")
print(df_encoded)
```

```python
# Using sklearn

from sklearn.preprocessing import OneHotEncoder
import pandas as pd

# Sample data
data = {'Color': ['Red', 'Blue', 'Green', 'Blue', 'Red']}
df = pd.DataFrame(data)

# Initialize the OneHotEncoder
encoder = OneHotEncoder(sparse_output = False)

# Fit and transform the data
encoded_data = encoder.fit_transform(df[['Color']])

# Convert the encoded data to a DataFrame with appropriate column names
encoded_df = pd.DataFrame(encoded_data, columns=encoder.categories_[0])

print("Original DataFrame:")
print(df)

print("\nOne-Hot Encoded DataFrame:")
print(encoded_df)
```
